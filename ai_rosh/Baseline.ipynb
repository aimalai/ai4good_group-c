```python
import polars as pl

print("TRAINING DISTRIBUTION")
train_distribution = (
    train_dataset.examples
    .group_by("label")
    .len()
    .sort("label")
    .with_columns(
        (pl.col("len") / len(train_dataset) * 100)
        .round(2)
        .alias("percentage")
    )
)
print(train_distribution)

print("\nVALIDATION DISTRIBUTION")
val_distribution = (
    val_dataset.examples
    .group_by("label")
    .len()
    .sort("label")
    .with_columns(
        (pl.col("len") / len(val_dataset) * 100)
        .round(2)
        .alias("percentage")
    )
)
print(val_distribution)
```

```python
import random
import numpy as np
import torch
import lightning as L
from pathlib import Path

# Experiment settings
SEED = 42
SUBSET_SIZES = [100, 1_000, 10_000, 100_000, 500_000]

# Reproducibility
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
L.seed_everything(SEED, workers=True)

# Load the original baseline configuration
config = load_config(CONFIG_PATH)

# Keep the original configuration, changing only the experiment seed
config["seed"] = SEED

# Build the original fold-0 datamodule
mapping = build_mapping(config)
datamodule = build_datamodule(config, mapping)
datamodule.setup("fit")

train_dataset = datamodule.train_set
val_dataset = datamodule.val_set

# Create one shuffled ordering, then take nested subsets
shuffled_examples = train_dataset.examples.sample(
    fraction=1.0,
    shuffle=True,
    seed=SEED,
)

training_subsets = {
    size: shuffled_examples.head(size)
    for size in SUBSET_SIZES
}

print("Experiment setup complete!")
print(f"Full training examples: {len(train_dataset.examples):,}")
print(f"Validation examples:    {len(val_dataset.examples):,}")
print(f"Subset sizes:           {list(training_subsets.keys())}")
```

```python
import time
import torch
import lightning as L

# Improve Tensor Core performance on the L40S
torch.set_float32_matmul_precision("high")

# Keep the model and training configuration unchanged
smoke_config = copy.deepcopy(config)
smoke_config["seed"] = SEED
smoke_config["trainer"]["max_epochs"] = 1

# Reduce DataLoader multiprocessing for the smoke test
smoke_config["data"]["num_workers"] = 0

# Rebuild a fresh datamodule
smoke_mapping = build_mapping(smoke_config)
smoke_dm = build_datamodule(smoke_config, smoke_mapping)
smoke_dm.setup("fit")

smoke_dm.train_set.examples = training_subsets[100].clone()

# Rebuild model
smoke_model = build_module(smoke_config, smoke_mapping)

# Safer smoke-test trainer
smoke_trainer = L.Trainer(
    max_epochs=1,
    accelerator="gpu",
    devices=1,
    precision="bf16-mixed",
    limit_train_batches=1.0,
    limit_val_batches=1,
    num_sanity_val_steps=0,
    logger=False,
    enable_checkpointing=False,
    enable_progress_bar=True,
)

start_time = time.time()

smoke_trainer.fit(smoke_model, datamodule=smoke_dm)

print(f"Smoke test completed in {(time.time() - start_time):.2f} seconds")
```

```python
f1_columns = [col for col in metrics_df.columns if "f1" in col.lower()]

print("F1-related metrics:")
for col in f1_columns:
    print(col)

print("\nRows containing F1 results:")
display(
    metrics_df.loc[
        metrics_df[f1_columns].notna().any(axis=1),
        ["epoch"] + f1_columns
    ].tail(5)
)
```

```python
horizons = [14, 30, 60, 365]

result_columns = {}

for horizon in horizons:
    result_columns[f"Binary F1 ({horizon}d)"] = (
        f"operational/binary_f1_{horizon}d"
    )
    result_columns[f"Macro F1 ({horizon}d)"] = (
        f"operational/macro_f1_{horizon}d"
    )

final_results = metrics_df[
    metrics_df[list(result_columns.values())].notna().any(axis=1)
].iloc[-1]

results_100 = pd.DataFrame({
    "Metric": list(result_columns.keys()),
    "Score": [
        final_results[col] for col in result_columns.values()
    ]
})

print("100-acquisition experiment — final epoch")
display(results_100)
```
